## 2. Download Model & Data
Download pre-trained models and data files from GitHub.

```python
import os, requests
BASE = "https://github.com/caizefan34/urban-mobility-ai/raw/master/space"
files = ["forecast_models.joblib","zones.csv","zones.json","travel_times.csv","predictions.parquet","app.py"]
os.makedirs("space", exist_ok=True)
for f in files:
    r = requests.get(f"{BASE}/{f}", stream=True)
    if r.status_code == 200:
        with open(f"space/{f}", "wb") as fh:
            [fh.write(chunk) for chunk in r.iter_content(8192)]
        print(f"  ✅ {f}")
    else:
        print(f"  ❌ {f} ({r.status_code})")
print("✅ All files downloaded")
```

## 3. Launch Demo
Run the Gradio app. A public URL will be generated.

```python
import subprocess, sys, time, os
os.chdir("space")
print("⏳ Launching Gradio app...")
p = subprocess.Popen([sys.executable,"app.py"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
time.sleep(8)
try:
    from google.colab.output import eval_js
    url = eval_js(f"google.colab.kernel.proxyPort(7860, {{'auth': false}})")
    print(f"\n🔗 **Open Demo: {url}**")
except:
    print("\n⚠️ Port 7860 accessible via Colab proxy")
print("⏹️  Press STOP in runtime menu when done")
```
